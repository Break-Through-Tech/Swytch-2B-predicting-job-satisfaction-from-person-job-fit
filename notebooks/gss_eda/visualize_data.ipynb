```python
df = df.dropna(subset=['SATJOB'])
```

```python
satjob_mapping = {
    1: 'Very Satisfied',
    2: 'Moderately Satisfied',
    3: 'A Little Dissatisfied',
    4: 'Very Dissatisfied'
}

df['SATJOB'] = df['SATJOB'].replace(satjob_mapping)
```

```python
sns.set_theme(style="whitegrid")

# Create a count plot for Job Satisfaction
plt.figure(figsize=(8, 5))
sns.countplot(
    data=df,
    x='SATJOB',
    order=[
        'Very Satisfied',
        'Moderately Satisfied',
        'A Little Dissatisfied',
        'Very Dissatisfied'
    ]
)
plt.title('Distribution of Job Satisfaction in GSS')
plt.ylabel('Number of Respondents')
plt.xlabel('Satisfaction Level')
plt.xticks(rotation=45)
plt.show()
```

```python
# Job satisfaction grouped by gender
plt.figure(figsize=(10, 6))
df = df.dropna(subset=['SEX']).copy()
df['SEX'] = df['SEX'].replace({1: 'Male', 2: 'Female'})

sns.countplot(
    data=df,
    x='SATJOB',
    hue='SEX',
    order=[
        'Very Satisfied',
        'Moderately Satisfied',
        'A Little Dissatisfied',
        'Very Dissatisfied'
    ]
)
plt.title('Job Satisfaction by Gender')
plt.legend(title='Gender')
plt.xticks(rotation=45)
plt.show()
```

```python
broad_features = [
    'SATJOB', # Target
    'AGE', 'EDUC', 'SEX', 'RACE', 'MARITAL', # Demographics
    'RINCOM16', 'INCOME16', 'CLASS',         # Wealth/Class
    'WRKSTAT', 'HRS1', 'UNION',              # Work characteristics
    'HEALTH', 'HAPPY', 'LIFE',               # Well-being
    'REGION', 'PARTYID'                      # Location/Ideology
]
```

```python
## Wider Net data frame
df_wide = df[broad_features].copy()
```

```python
# filter / clean wide data 
df_wide = df_wide[~df_wide['SATJOB'].isin([0, 8, 9])]
df_wide = df_wide.dropna(subset=['SATJOB'])
```

```python
print("Max Age:", df_wide['AGE'].max())
print("Max Hours Worked:", df_wide['HRS1'].max())

print("\nTop 5 highest AGE values and their counts:")
print(df_wide['AGE'].value_counts().sort_index().tail())
```

```python
# Impute missing values 
print(df_wide.dtypes)
```

```python
## Impute missing values 
numerical_cols = ['AGE', 'EDUC', 'RACE', 'MARITAL', 'RINCOM16', 'INCOME16', 'CLASS', 'WRKSTAT', 'HRS1', 'UNION', 'HEALTH', 'HAPPY', 'LIFE', 'REGION', 'PARTYID']
categorical_cols = ['SEX']

for col in numerical_cols:
    if col in df_wide.columns:
        df_wide[col] = df_wide[col].fillna(df_wide[col].median())

for col in categorical_cols:
    if col in df_wide.columns:
        df_wide[col] = df_wide[col].fillna(df_wide[col].mode()[0])
print(df_wide.isnull().sum())
```

```python
# Cast to string for one-hot-encoding 
for col in categorical_cols:
    if col in df_wide.columns:
        df_wide[col] = df_wide[col].astype(str)
```
